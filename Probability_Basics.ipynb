Question 1: A die is rolled. What is the probability of getting:
(a) An even number

(b) A number greater than 4

Sol - (a) Probability of getting an even number:
The even numbers are 2, 4, and 6. Therefore, the probability is 3/6 = 1/2.

(b) Probability of getting a number greater than 4:
The numbers greater than 4 are 5 and 6. Therefore, the probability is 2/6 = 1/3.





Question 2: In a class of 50 students:
- 20 like Mathematics (M)
- 15 like Science (S)

 - 5 like both subjects

 - What is the probability that a student chosen at random likes Mathematics or Science?

Sol - In a class of 50 students, 20 like Mathematics, 15 like Science, and 5 like both subjects.
- Using the formula:
- P(M or S) = [M + S − (M and S)] / Total students
- = (20 + 15 − 5) / 50
- = 30 / 50
- = 0.6
- Therefore, the probability that a randomly chosen student likes Mathematics or Science is 0.6 or 60%.

Question 3: A bag has 3 red and 2 blue balls. If one ball is drawn randomly and is red, what is the probability that the next ball is also red (without replacement)?

Sol - The bag has 3 red and 2 blue balls, making a total of 5 balls. Since the first ball drawn is red, 2 red and 2 blue balls remain, making a total of 4 balls.
- Therefore, the probability that the next ball is also red is:
- Probability = Number of remaining red balls / Total remaining balls
- = 2 / 4
- = 1/2 = 50%
- So, the probability of drawing another red ball is 1/2 or 50%.

Question 4: The population of a school is divided into 60% boys and 40% girls. If you want equal representation of both genders in the sample, which method should you use: Simple Random Sampling or Stratified Sampling? Why?

Sol - The appropriate method is Stratified Sampling.
- In this method, the population is divided into two groups (strata): boys (60%) and girls (40%). Samples are then selected from each group to ensure proper representation.
- Since the goal is to have equal representation of boys and girls, stratified sampling is better than simple random sampling because it ensures that both groups are included in the desired proportions.
- Answer: Use Stratified Sampling because it allows the researcher to select students separately from each gender group and ensure equal representation.

Question 5: The average height of 1000 students = 160 cm. A sample of 100 students shows an average height = 158 cm.
- Find the sampling error.

Sol - Population average height = 160 cm
- Sample average height = 158 cm
- Formula:
- Sampling Error = Sample Mean − Population Mean
- = 158 − 160
- = −2 cm
- Therefore, the sampling error is −2 cm, meaning the sample average is 2 cm lower than the actual population average.

Question 6: The population mean salary is ₹50,000 with σ = ₹5,000. If we take a sample of 100 employees, what is the standard error of the mean (SEM)?

Sol - Given:
- Population standard deviation (σ) = ₹5,000
- Sample size (n) = 100
- Formula:
- Standard Error (SEM) = σ / √n
- = 5,000 / √100
- = 5,000 / 10
- = ₹500
- Therefore, the standard error of the mean is ₹500.

Sol - There are 100 students.
- Students who like Cricket = 40

Question 8: From a deck of 52 cards, two cards are drawn without replacement. What is the probability that both are Aces?

Sol - A standard deck has 52 cards, including 4 Aces.
Since the cards are drawn without replacement, after drawing one Ace, there are 3 Aces left out of 51 cards.
- Probability of first card being an Ace:
- 4 / 52 = 1 / 13
- Probability of second card being an Ace:
- 3 / 51 = 1 / 17
- Therefore:
- P(Both Aces) = (4/52) × (3/51)
- = 12 / 2652
- = 1/221
- Therefore, the probability of drawing two Aces is 1/221, or approximately 0.45%.

Question 9: A factory produces bulbs with 2% defective rate. If 5 bulbs are chosen at random, what is the probability that all are non-defective?

Sol - The defective rate is 2%, so the non-defective rate is:
- 100% − 2% = 98% = 0.98
- For 5 bulbs, the probability that all are non-defective is:
- P(All non-defective) = (0.98)^5
- = 0.9039
- Therefore, the probability that all 5 bulbs are non-defective is approximately 0.9039 or 90.39%.

Question 10: Differentiate between discrete and continuous random variables with examples.

Sol - A discrete random variable takes specific, countable values. It usually represents things that can be counted.
- Example: Number of students in a class, number of defective bulbs, or number of heads when tossing coins.
- A continuous random variable can take any value within a range. It usually represents measurements.
- Example: Height of a student, weight of a person, temperature, or time taken to complete a task.
- Discrete Random Variable      	- Continuous Random Variable
- Takes countable values        	- Takes any value within a range
- Usually represents counts     	- Usually represents measurements
- Example: Number of students   	- Example: Height of students


- In short: Discrete variables are counted, while continuous variables are measured.
